# Project Overview: Business Survey Data Analysis

This project focuses on the **cleaning, management, and analysis of business survey data collected in Migori during the 2025/2026 field study**. The dataset was gathered through a structured questionnaire administered by enumerators, each tasked with collecting responses from three businesses within the study area. The survey aimed to capture information on the **characteristics of local enterprises, ownership structures, employment size, operational history, and the challenges faced by businesses**.  

The workflow of this project begins with **data preparation using SQL**, where the raw survey dataset is cleaned and structured for analysis. This process involves standardizing column names, correcting inconsistent responses, handling missing values, verifying enumerator submissions, and organizing categorical variables into a format suitable for analysis. Care is taken to preserve all original observations since each record represents a unique business surveyed in the field.  

Once cleaned, the dataset is transformed through **aggregation queries** to produce summary statistics such as the distribution of businesses by sector, ownership category, gender of business owners, years of operation, and the major constraints affecting enterprises. These aggregations help convert raw questionnaire responses into meaningful analytical indicators.  

The cleaned and structured data is then connected to **Microsoft Power BI**, where interactive dashboards are developed to visualize the results. The dashboards provide insights into the **local economic structure, business diversity, employment patterns, and operational challenges faced by small enterprises** in the study area. Visualizations such as bar charts, pie charts, and key performance indicators enable clear communication of patterns observed in the data.  

Overall, this project demonstrates how **field-collected survey data can be transformed into actionable insights through data cleaning, SQL-based aggregation, and data visualization tools**. The final dashboard serves as a decision-support resource that can assist planners, researchers, and local stakeholders in understanding the **business environment and identifying areas for economic support and policy intervention**.

### Spatial Analysis of Ownership Structures
To determine if the vibrancy of Lower Marindi (CBD) encourages collaborative ownership (Partnerships) compared to the individualistic nature of Wasweta 1.

```python
%%sql
CREATE OR REPLACE VIEW ownership_spatial_analysis AS
SELECT 
    Place,
    Ownership_Category,
    COUNT(*) AS total_count,
    -- Checking how many of these actually managed to scale
    SUM(CASE WHEN Employee_Size IN ('6-10 employees', 'Above 10', '6-10', 'Above 10') THEN 1 ELSE 0 END) AS expanded_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM business_data WHERE Place = b.Place), 2) AS percentage_within_place
FROM business_data b
WHERE Place IN ('Lower Marindi', 'Wasweta 1')
GROUP BY Place, Ownership_Category
ORDER BY Place, total_count DESC;
SELECT*
FROM ownership_spatial_analysis;
```

### Sector Specialization & The Expansion Gap
To identify if the CBD Productivity Bonus in Lower Marindi is driven by higher-value sectors (Services/Wholesale) compared to the retail-heavy outskirts of Wasweta 1.

```python
%%sql
CREATE OR REPLACE VIEW sector_specialization_expansion AS
SELECT 
    b.Place,
    s.sector,
    COUNT(DISTINCT b.id) AS unique_businesses,
    # Count how many of these unique businesses scaled
    COUNT(DISTINCT CASE WHEN b.Employee_Size IN ('6-10 employees', 'Above 10', '6-10', 'Above 10') THEN b.id END) AS expanded_count,
    # Percentage of the unique 66-record sample
    ROUND(COUNT(DISTINCT b.id) * 100.0 / (SELECT COUNT(DISTINCT id) FROM business_data WHERE Place = b.Place), 2) AS percent_of_local_economy
FROM business_data b
JOIN business_sector_activity s ON b.id = s.business_id
WHERE b.Place IN ('Lower Marindi', 'Wasweta 1')
  AND s.sector <> 'None'
GROUP BY b.Place, s.sector
ORDER BY b.Place, unique_businesses DESC;
SELECT*
FROM sector_specialization_expansion;
```

```python
%%sql
CREATE OR REPLACE VIEW sector_specialization_aggregation AS
SELECT 
    Place,
    CASE 
        WHEN sector LIKE '%General Traders%' THEN 'General Traders'
        WHEN sector LIKE '%Informal Sector%' THEN 'Informal Sector'
        WHEN sector LIKE '%Professional%' THEN 'Professional/Financial'
        ELSE 'Other Specialized'
    END AS Sector_Grouped,
    SUM(unique_businesses) AS Total_Businesses,
    SUM(expanded_count) AS Total_Expanded,
    ROUND(AVG(percent_of_local_economy), 1) AS Avg_Market_Share_Percent
FROM sector_specialization_expansion
WHERE Place IN ('Lower Marindi', 'Wasweta 1')
GROUP BY Place, Sector_Grouped
ORDER BY Place, Total_Businesses DESC;
SELECT* 
FROM sector_specialization_aggregation;
```

### Specialized Engine vs. The Retail
#### Quality over Quantity: The Other Specialized Factor
In planning, we often look for the **economic engine** of a town. The data shows that while Retail is more common, **the Other Specialized category (Industrial, Transport, Communications) is the real job creator in the CBD.**

**Lower Marindi**: Out of 12 specialized businesses, 5 have expanded. That is a 41.7% scaling rate.

**Wasweta 1**: Out of 4 specialized businesses, 0 have expanded. That is a 0% scaling rate.

In the CBD, a specialized business is nearly 42% likely to expand, whereas in the periphery, these same sectors struggle to break the 5-employee ceiling. Location acts as a catalyst for high-order services.

#### The Retail Monopoly(Wasweta 1)
Wasweta 1 is essentially "locked" into General Trading, which accounts for 73.3% of its local economy. This is a **sectoral over-specialization**.
Because almost everyone is selling similar goods to the same limited local market, the "Expansion Velocity" is incredibly low.

Even though Wasweta 1 has 22 traders (more than Lower Marindi’s 18), only 3 have managed to scale. This represents a modest 13.6% expansion rate for retail in the outskirts compared to the high-growth specialized sectors in the heart of town.

### Vibrancy = Diversity (The Agglomeration Economy)
The vibrancy of Migori’s CBD is more than just a feeling; it’s visible in the "Other Specialized" count.

**Lower Marindi has 3x more specialized businesses (Industrial, Transport, Communication) than Wasweta 1.**

This diversity is the foundation of an Agglomeration Economy. In the CBD, businesses don't just exist side by side; they support each other. A transport firm helps a wholesaler, who in turn hires a professional service. This synergy is what allows these businesses to grow, whereas the "Retail Monopoly" in Wasweta 1 leads to stagnation.

```python
%%sql 
CREATE OR REPLACE VIEW gender_barrier_summary AS
SELECT 
    Place,
    Gender,
    Education,
    ROUND(AVG(CASE 
        WHEN Years_Operated = 'Less than 1 yr' THEN 0.5
        WHEN Years_Operated = '1-5 yrs' THEN 3
        WHEN Years_Operated = '6-10 yrs' THEN 8
        WHEN Years_Operated = 'More than 10 yrs' THEN 12
        ELSE 0 
    END), 1) AS avg_years_tenure,
    # Top Hurdle Counts for ALL 7 Categories
    SUM(CASE WHEN High_cost = '1' THEN 1 ELSE 0 END) AS hurdle_high_cost,
    SUM(CASE WHEN Taxes = '1' THEN 1 ELSE 0 END) AS hurdle_taxes,
    SUM(CASE WHEN Acces_finance = '1' THEN 1 ELSE 0 END) AS hurdle_finance,
    SUM(CASE WHEN Access_premises = '1' THEN 1 ELSE 0 END) AS hurdle_premises,
    SUM(CASE WHEN Infrastructure_gaps = '1' THEN 1 ELSE 0 END) AS hurdle_infra,
    SUM(CASE WHEN Sanitation = '1' THEN 1 ELSE 0 END) AS hurdle_sanitation,
    SUM(CASE WHEN Competition = '1' THEN 1 ELSE 0 END) AS hurdle_competition,
    # Scaling success
    SUM(CASE WHEN Employee_Size IN ('6-10 employees', 'Above 10', '6-10', 'Above 10') THEN 1 ELSE 0 END) AS expansion_count
FROM business_data
GROUP BY Place, Gender, Education;

SELECT * FROM gender_barrier_summary;
```

### Sectoral Hurdle Analysis
To determine if high-growth specialized sectors face different physical or regulatory barriers than the retail sector.

```python
%%sql
CREATE OR REPLACE VIEW sector_hurdle_analysis_v3 AS
SELECT 
    CASE 
        WHEN retail_services IS NOT NULL AND retail_services != 'N/A' AND retail_services != '' THEN 'Retail & General Trade'
        WHEN Industrial_Factories IS NOT NULL AND Industrial_Factories != 'N/A' AND Industrial_Factories != '' THEN 'Industrial/Manufacturing'
        WHEN (Transport IS NOT NULL AND Transport != 'N/A' AND Transport != '') 
          OR (Storage IS NOT NULL AND Storage != 'N/A' AND Storage != '') 
          OR (Communications IS NOT NULL AND Communications != 'N/A' AND Communications != '') THEN 'Logistics & Tech'
        WHEN Informal_Sector IS NOT NULL AND Informal_Sector != 'N/A' AND Informal_Sector != '' THEN 'Informal/Small Scale'
        WHEN (Accomodation_Catering IS NOT NULL AND Accomodation_Catering != 'N/A' AND Accomodation_Catering != '') 
          OR (Entertainment IS NOT NULL AND Entertainment != 'N/A' AND Entertainment != '') THEN 'Hospitality & Leisure'
        WHEN (PTF_Services IS NOT NULL AND PTF_Services != 'N/A' AND PTF_Services != '') 
          OR (Health IS NOT NULL AND Health != 'N/A' AND Health != '') 
          OR (Education_Institution IS NOT NULL AND Education_Institution != 'N/A' AND Education_Institution != '') THEN 'Professional & Social Services'
        WHEN AFN_Activities IS NOT NULL AND AFN_Activities != 'N/A' AND AFN_Activities != '' THEN 'Primary Sector (Agri/Forest/Res)'
        ELSE 'Unclassified'
    END AS Detailed_Sector_Group,
    Place,
    COUNT(*) AS total_businesses,
    # Top Hurdle Counts
    SUM(CASE WHEN High_cost = '1' THEN 1 ELSE 0 END) AS hurdle_high_cost,
    SUM(CASE WHEN Taxes = '1' THEN 1 ELSE 0 END) AS hurdle_taxes,
    SUM(CASE WHEN Acces_finance = '1' THEN 1 ELSE 0 END) AS hurdle_finance,
    SUM(CASE WHEN Access_premises = '1' THEN 1 ELSE 0 END) AS hurdle_premises,
    SUM(CASE WHEN Infrastructure_gaps = '1' THEN 1 ELSE 0 END) AS hurdle_infra,
    SUM(CASE WHEN Sanitation = '1' THEN 1 ELSE 0 END) AS hurdle_sanitation,
    SUM(CASE WHEN Competition = '1' THEN 1 ELSE 0 END) AS hurdle_competition,
    # Scaling success
    SUM(CASE WHEN Employee_Size IN ('6-10 employees', 'Above 10', '6-10', 'Above 10') THEN 1 ELSE 0 END) AS expansion_count
FROM business_data
GROUP BY Detailed_Sector_Group, Place;

SELECT * FROM sector_hurdle_analysis_v3;
```

### Sectoral Barrier Analysis: Growth vs. Stagnation
1. The High-Growth Specialized Sectors (Logistics, Tech, & Manufacturing)
The Logistics & Tech group in Lower Marindi is the **Economic Engine** with a 75% expansion rate (3 out of 4).

Unlike Retail, these specialized businesses are the only ones citing Infrastructure Gaps (1) and High Costs (2) as their primary friction points.
These sectors are ready to scale, but they are physically limited. Their growth depends on the county’s ability to provide stable **Backend** infrastructure (power, roads, and digital connectivity).

3. The Low-Growth Retail
In Wasweta 1, Retail & General Trade is the largest employer (22 businesses) but has a low expansion rate of only 13%.  
This sector is uniquely crushed by Competition (7) and Sanitation (3).  

Retail is in a Survival Loop. Because there are too many identical shops in the same space, profits are thin, and expansion is nearly impossible. Planning for Wasweta 1 shouldn't just be about more space for shops, but about zoning for variety to reduce this cut-throat competition.

3. The Institutional Burden (Taxes)
Taxes are almost exclusively a CBD problem. In Lower Marindi, Retail (7) and Logistics (1) cite taxes heavily. In Wasweta 1, taxes almost disappear as a no.1 hurdle.

There is a **Formalization Penalty** in the CBD. Businesses that grow large enough to be noticed by the county are being taxed at a rate that may be hindering their next stage of hiring.

### Other Issues Analysis

```python
%%sql
CREATE OR REPLACE VIEW business_stagnation_voices AS
SELECT 
    id, -- Added ID for easier reference
    Place,
    Detailed_Sector_Group,
    `Other Issues`,
    Access_premises
FROM (
    SELECT 
        id,
        Place,
        CASE 
            WHEN retail_services IS NOT NULL AND retail_services NOT IN ('', 'N/A', '0') THEN 'Retail & General Trade'
            WHEN Industrial_Factories IS NOT NULL AND Industrial_Factories NOT IN ('', 'N/A', '0') THEN 'Industrial/Manufacturing'
            WHEN (Transport IS NOT NULL AND Transport NOT IN ('', 'N/A', '0')) 
              OR (Storage IS NOT NULL AND Storage NOT IN ('', 'N/A', '0')) 
              OR (Communications IS NOT NULL AND Communications NOT IN ('', 'N/A', '0')) THEN 'Logistics & Tech'
            WHEN Informal_Sector IS NOT NULL AND Informal_Sector NOT IN ('', 'N/A', '0') THEN 'Informal/Small Scale'
            WHEN (Accomodation_Catering IS NOT NULL AND Accomodation_Catering NOT IN ('', 'N/A', '0')) 
              OR (Entertainment IS NOT NULL AND Entertainment NOT IN ('', 'N/A', '0')) THEN 'Hospitality & Leisure'
            WHEN (PTF_Services IS NOT NULL AND PTF_Services NOT IN ('', 'N/A', '0')) 
              OR (Health IS NOT NULL AND Health NOT IN ('', 'N/A', '0')) 
              OR (Education_Institution IS NOT NULL AND Education_Institution NOT IN ('', 'N/A', '0')) THEN 'Professional & Social Services'
            WHEN AFN_Activities IS NOT NULL AND AFN_Activities NOT IN ('', 'N/A', '0') THEN 'Primary Sector (Agri/Forest/Res)'
            ELSE 'Unclassified'
        END AS Detailed_Sector_Group,
        `Other Issues`,
        Access_premises,
        Employee_Size
    FROM business_data
) AS subquery
WHERE 
    -- Condition 1: Include businesses that haven't expanded
    (Employee_Size NOT IN ('6-10 employees', 'Above 10', '6-10', 'Above 10')
     OR id = 41) -- Exception for the specific scaled business with issues
    AND 
    (
      (`Other Issues` IS NOT NULL AND `Other Issues` NOT IN ('N/A', '', 'None', '0')) 
      OR 
      (Access_premises IS NOT NULL AND Access_premises NOT IN ('N/A', '', 'None', '0'))
    );

SELECT* 
FROM business_stagnation_voices;
```

### Thematic Categorization(PROBLEM TREE)
Why certain businesses are stuck (stagnation) while others are struggling to maintain their scale.


```python
%%sql
SELECT 
    Category,
    GROUP_CONCAT(id ORDER BY id SEPARATOR ', ') AS Specific_IDs,
    CASE 
        WHEN Category = 'Environmental Design' THEN 'Safety & Flow: "Insecurity" and "Low flow" suggest poor street lighting and a lack of pedestrian-friendly paths.'
        WHEN Category = 'Accessibility Crisis' THEN 'Premises Connectivity: Scores of 1 or 2 in Access_premises indicate physical difficulty or unappealing entry for customers.'
        WHEN Category = 'Market Volatility' THEN 'Economic Resilience: "Haggling" and "Volatility" suggest a lack of a stable, formalized market environment.'
        ELSE 'General Operational Constraints'
    END AS Planning_Issue_Identified
FROM (
    SELECT 
        id,
        CASE 
            -- Identifying Environmental Design (Insecurity or Low Flow)
            WHEN `Other Issues` LIKE '%Insecurity%' OR `Other Issues` LIKE '%Low flow%' THEN 'Environmental Design'
            
            -- Identifying Accessibility Crisis (Access_premises score of 1 or 2)
            WHEN Access_premises IN ('1', '2') THEN 'Accessibility Crisis'
            
            -- Identifying Market Volatility (Haggling, Volatility, or specific Depts issues)
            WHEN `Other Issues` LIKE '%Haggling%' OR `Other Issues` LIKE '%Volatility%' OR `Other Issues` LIKE '%depts%' THEN 'Market Volatility'
            
            ELSE 'Other Issues'
        END AS Category
    FROM business_data
    WHERE 
        -- Ensuring we only categorize rows with actual qualitative data
        ((`Other Issues` IS NOT NULL AND `Other Issues` NOT IN ('N/A', '', 'None', '0')) 
        OR 
        (Access_premises IS NOT NULL AND Access_premises NOT IN ('N/A', '', 'None', '0')))
) AS Categorized_Data
GROUP BY Category;
```

### Planning Interventions Matrix

```python
%%sql
CREATE OR REPLACE VIEW planning_intervention_matrix AS
SELECT 
    Category,
    Planning_Issue_Identified,
    CASE 
        WHEN Category = 'Environmental Design' THEN '1. Install high-mast street lighting in Lower Marindi to curb insecurity. 2. Redesign the A1 corridor to prioritize pedestrian walkways and safe "stop-over" points.'
        WHEN Category = 'Accessibility Crisis' THEN '1. Implement a "Frontage Improvement Program" to clear encroachments. 2. Pave drainage-sensitive access points in Wasweta 1 to ensure year-round business access.'
        WHEN Category = 'Market Volatility' THEN '1. Establish a Migori Business Resource Center to provide market data and price guidance. 2. Implement zoning for "Sector Hubs" to reduce retail oversaturation.'
        ELSE 'General Urban Management Improvements'
    END AS Proposed_Planning_Interventions
FROM (
    SELECT 
        CASE 
            WHEN `Other Issues` LIKE '%Insecurity%' OR `Other Issues` LIKE '%Low flow%' THEN 'Environmental Design'
            WHEN Access_premises IN ('1', '2') THEN 'Accessibility Crisis'
            WHEN `Other Issues` LIKE '%Haggling%' OR `Other Issues` LIKE '%Volatility%' OR `Other Issues` LIKE '%depts%' THEN 'Market Volatility'
            ELSE 'Other'
        END AS Category,
        CASE 
            WHEN `Other Issues` LIKE '%Insecurity%' OR `Other Issues` LIKE '%Low flow%' THEN 'Safety & Flow issues'
            WHEN Access_premises IN ('1', '2') THEN 'Physical Premises Connectivity'
            WHEN `Other Issues` LIKE '%Haggling%' OR `Other Issues` LIKE '%Volatility%' OR `Other Issues` LIKE '%depts%' THEN 'Economic Resilience'
            ELSE 'General Constraints'
        END AS Planning_Issue_Identified
    FROM business_data
    WHERE (`Other Issues` IS NOT NULL AND `Other Issues` NOT IN ('N/A', '', 'None')) 
       OR (Access_premises IS NOT NULL AND Access_premises NOT IN ('N/A', '', 'None'))
) AS Sub
GROUP BY Category, Planning_Issue_Identified;

SELECT * FROM planning_intervention_matrix;
```

This study conducted a comprehensive spatial-economic analysis of Migori Town, utilizing data from over 60 business entities across three key character areas: Lower Marindi (CBD), Wasweta 1, and Oruba. ### Key Findings

The analysis reveals a **growth ceiling** effect. While the Retail and General Trade sector dominates the landscape by volume (particularly in Wasweta 1), it suffers from the lowest expansion rates (approx.13%) due to extreme market saturation and hyper-competition. Conversely, specialized sectors like Logistics, Tech, and Manufacturing in Lower Marindi demonstrate a higher propensity for growth (up to 75% expansion) but are physically constrained by a lack of specialized infrastructure and high operating costs.

Furthermore, the data suggests that Employees provide a more critical assessment of urban failure (citing insecurity and access) than Owners, who tend to focus on financial hurdles like taxes and rent.

3. The Accessibility & Security Crisis
The Human Voice(Other Issues) highlighted a major spatial failure:  

**Connectivity**:Over 70% of stagnating businesses rated their physical accessibility at a 2/5 or lower.  

**Environmental Safety**: In Lower Marindi, even businesses with Above 10 employees reported **Low Customer Flow** and **Insecurity** as primary threats to their sustainability. This indicates that poor urban design (lighting and pedestrian paths) is actively devaluing the CBD’s economic potential.  

### Planning Recommendations
To revitalize Migori Town’s economic landscape, the study proposes a Data-Driven Intervention Matrix:

1. Physical Infrastructure: Immediate paving and drainage improvement in Wasweta 1 to break the "Accessibility Crisis."

2. Public Safety: Implementation of high-mast lighting and pedestrian-centric "A1 Corridor" redesign in Lower Marindi to facilitate 24-hour economy flow.

3. Economic Policy: Transitioning from general licensing to Sector-Specific Zoning, encouraging the formation of "Specialized Hubs" to reduce retail oversaturation.

The study concludes that Migori Town’s stagnation is not merely a financial issue but a spatial one. The evidence shows that businesses are ready to scale, but the current urban fabric characterized by poor access, insecurity, and chaotic zoning acts as a physical barrier to expansion.
